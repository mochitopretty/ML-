<h2 style="color:Fuchsia; font-family:; text-align:center;font-size:15px">
  Caso 6: Evitar falsas alarmas ❁۪ ｡˚
</h2>
Un sistema automático monitorea una instalación experimental. La mayor parte del tiempo todo funciona normalmente. Cada vez que el modelo detecta una posible falla, el experimento debe detenerse para revisar el sistema. Pregunta: ¿Qué modelo elegirían si detener el experimento por una falsa alarma tiene un costo alto? ¿Qué tipo de error están tratando de minimizar?
<h2 style="color:Fuchsia; font-family:; text-align:center;font-size:15px">
   ❁
</h2>
Si detener el experimento por una falsa alarma tiene un costo alto, se podría escoger el modelo B dado que tiene un accuracy de 96,9%, puede ser útil puesto que representa el porcentaje de clasificaciones correctas. Además, aquello se puede relacionar con el modelo que tenga menos FP (25 del B con respecto a 120 del modelo A (esto es tomando en cuenta la matriz de confusión, TN FP, y abajo FN TP). Lo que se quiere minimizar es que el sistema sea falso positivo, es decir, las falsas alarmas (negativos) clasificados incorrectamente, dado que el sistema funciona bien. Al comparar entre los valores, obtenidos desde las métricas de la matriz de confusión, se obtuvo que el modelo A, tiene un accuracy de 87,7%, una precisión (pureza) de 28,14%, un recall (completitud) de 94% y un F1 (para encontrar un equilibrio entre precisión y recall) de 43,3%. Por otra parte, para el modelo B se tiene tiene un accuracy de 96,9%, una precisión (pureza) de 63,76%%, un recall (completitud) de 88% y un F1 (para encontrar un equilibrio entre precisión y recall) de 73,94%.
